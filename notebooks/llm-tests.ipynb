# **Helper Functions**


### EXIF data

````python
from PIL import Image
from PIL.ExifTags import TAGS, GPSTAGS


def convert_to_decimal(coords, ref):
    degrees = float(coords[0])
    minutes = float(coords[1])
    seconds = float(coords[2])

    decimal = degrees + (minutes / 60) + (seconds / 3600)

    if ref in ["S", "W"]:
        decimal = -decimal

    return float(decimal)


def extract_metadata(file_path):

    with Image.open(file_path) as img:

        exif_data = img.getexif()

        readable_exif = {}

        for tag_id, value in exif_data.items():
            tag_name = TAGS.get(tag_id, tag_id)
            readable_exif[tag_name] = value

        date_time = (
            readable_exif.get("DateTimeOriginal")
            or readable_exif.get("DateTimeDigitized")
            or readable_exif.get("DateTime")
        )

        latitude = None
        longitude = None
        altitude = None

        try:
            gps_ifd = exif_data.get_ifd(0x8825)

            readable_gps = {}

            for tag_id, value in gps_ifd.items():
                tag_name = GPSTAGS.get(tag_id, tag_id)
                readable_gps[tag_name] = value

            if (
                "GPSLatitude" in readable_gps
                and "GPSLatitudeRef" in readable_gps
                and "GPSLongitude" in readable_gps
                and "GPSLongitudeRef" in readable_gps
            ):

                latitude = convert_to_decimal(
                    readable_gps["GPSLatitude"],
                    readable_gps["GPSLatitudeRef"]
                )

                longitude = convert_to_decimal(
                    readable_gps["GPSLongitude"],
                    readable_gps["GPSLongitudeRef"]
                )

            if "GPSAltitude" in readable_gps:
                altitude = float(readable_gps["GPSAltitude"])

        except Exception as e:
            print("GPS extraction error:", e)

    return {
        "Date_Time": date_time,
        "Latitude": latitude,
        "Longitude": longitude,
        "Altitude": altitude
    }
````

### Location matching

````python
from math import radians, sin, cos, sqrt, atan2


def haversine(lat1, lon1, lat2, lon2):

    R = 6371  # Earth radius in kilometres

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        sin(dlat / 2) ** 2
        + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    )

    c = 2 * atan2(
        sqrt(a),
        sqrt(1 - a)
    )

    return R * c


def find_nearest_location(latitude, longitude, beach_lookup):

    if pd.isna(latitude) or pd.isna(longitude):
        return {
            "Location_Name": None,
            "Council": None,
            "Location_Distance_km": None,
            "Location_Source": "No GPS"
        }

    distances = []

    for _, beach in beach_lookup.iterrows():

        distance = haversine(
            latitude,
            longitude,
            beach["LATITUDE"],
            beach["LONGITUDE"]
        )

        distances.append(distance)

    nearest_index = distances.index(min(distances))
    nearest = beach_lookup.iloc[nearest_index]

    return {
        "Location_Name": nearest["BEACH_NAME"],
        "Council": nearest["COUNCIL"],
        "Location_Distance_km": min(distances),
        "Location_Source": "EXIF GPS + beach reference"
    }
````

### Temporal classficiation

````python
def get_season(month):
    if month in [12, 1, 2]:
        return "Summer"
    elif month in [3, 4, 5]:
        return "Autumn"
    elif month in [6, 7, 8]:
        return "Winter"
    elif month in [9, 10, 11]:
        return "Spring"
    else:
        return None


def get_sampling_period(row):
    year = int(row["Year"])
    month = int(row["Month"])
    season = row["Season"]

    if season == "Summer":
        if month == 12:
            return f"Summer_{year}_{str(year + 1)[-2:]}"
        else:
            return f"Summer_{year - 1}_{str(year)[-2:]}"

    return f"{season}_{year}"
````
